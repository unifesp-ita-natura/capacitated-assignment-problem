```python
import numpy as np
```

```python
MAX_SECTOR_DAILY_DEMAND = 100
MIN_SECTOR_DAILY_DEMAND = 80
N_SECTORS = 80
SEED = 42
CYCLE_LENGTHS = [20, 21, 20, 24]
```

```python
rng = np.random.default_rng(SEED)
```

```python
cycle_demand = {}
for cycle_id, cycle_length in enumerate(CYCLE_LENGTHS):
    cycle_sector_daily_demands = [
        rng.uniform(MIN_SECTOR_DAILY_DEMAND, MAX_SECTOR_DAILY_DEMAND, size=cycle_length)
        for _ in range(N_SECTORS)
    ]
    cycle_demand[cycle_id] = cycle_sector_daily_demands
```

```python
# pretend sectors start on day sector_id % 15
```
