```python
import pandas as pd
```

```python
zone_vs_setor_df = pd.read_excel("Bases_setores_cd.xlsx", sheet_name = "Setor vs Zoneamento")
```

```python
zone_vs_setor_df
```

```python
re_por_zoneamento_df = (
    zone_vs_setor_df
    .groupby("CD_ZONEAMENTO", as_index=False)
    .agg(
        QTD_RE=("CD_RE", "nunique")
    )
    .sort_values("QTD_RE", ascending=False)
)

re_por_zoneamento_df
```

```python
zoneamentos_por_setor_df = (
    zone_vs_setor_df
    .groupby("CD_SETOR", as_index=False)
    .agg(
        QTD_ZONEAMENTOS=("CD_ZONEAMENTO", "nunique")
    )
    .sort_values("QTD_ZONEAMENTOS", ascending=False)
)

zoneamentos_por_setor_df
```

```python
setor_por_zoneamentos_df = (
    zone_vs_setor_df
    .groupby("CD_ZONEAMENTO", as_index=False)
    .agg(
        QTD_SETORES=("CD_SETOR", "nunique")
    )
    .sort_values("QTD_SETORES", ascending=False)
)

setor_por_zoneamentos_df
```

```python
gvs_por_setor_df = (
    zone_vs_setor_df
    .groupby("CD_SETOR", as_index=False)
    .agg(
        QTD_GVS=("CD_GV", "nunique")
    )
    .sort_values("QTD_GVS", ascending=False)
)

gvs_por_setor_df
```

```python
gerencias_por_setor_df = (
    zone_vs_setor_df
    .groupby("CD_SETOR", as_index=False)
    .agg(
        QTD_GERENCIAS_VENDAS=("CD_GV", "nunique")
    )
    .sort_values("QTD_GERENCIAS_VENDAS", ascending=False)
)

gerencias_por_setor_df
```

```python
# Quantidade de GRUPOS distintos por GV
grupos_por_gv_df = (
    zone_vs_setor_df
    .groupby("CD_GV", as_index=False)
    .agg(
        QTD_GRUPOS=("CD_GRUPO", "nunique")
    )
    .sort_values("QTD_GRUPOS", ascending=False)
)

grupos_por_gv_df
```

```python
# Quantidade de GVs distintos por GRUPO
gvs_por_grupo_df = (
    zone_vs_setor_df
    .groupby("CD_GRUPO", as_index=False)
    .agg(
        QTD_GVS=("CD_GV", "nunique")
    )
    .sort_values("QTD_GVS", ascending=False)
)

gvs_por_grupo_df
```

```python
setores_por_grupo_df = (
    zone_vs_setor_df
    .groupby("CD_GRUPO", as_index=False)
    .agg(
        QTD_SETORES=("CD_SETOR", "nunique")
    )
    .sort_values("QTD_SETORES", ascending=False)
)

setores_por_grupo_df
```

```python
grupos_por_setor_df = (
    zone_vs_setor_df
    .groupby("CD_SETOR", as_index=False)
    .agg(
        QTD_GRUPOS=("CD_GRUPO", "nunique")
    )
    .sort_values("QTD_GRUPOS", ascending=False)
)

grupos_por_setor_df
```

```python
# Quantidade de zoneamentos distintos por grupo
zoneamentos_por_grupo_df = (
    zone_vs_setor_df
    .groupby("CD_GRUPO", as_index=False)
    .agg(
        QTD_ZONEAMENTOS=("CD_ZONEAMENTO", "nunique")
    )
    .sort_values("QTD_ZONEAMENTOS", ascending=False)
)

zoneamentos_por_grupo_df
```

```python
# Quantidade de grupos distintos por zoneamento
grupos_por_zoneamento_df = (
    zone_vs_setor_df
    .groupby("CD_ZONEAMENTO", as_index=False)
    .agg(
        QTD_GRUPOS=("CD_GRUPO", "nunique")
    )
    .sort_values("QTD_GRUPOS", ascending=False)
)

grupos_por_zoneamento_df
```

```python
# Quantidade de REs distintos por GV
res_por_gv_df = (
    zone_vs_setor_df
    .groupby("CD_GV", as_index=False)
    .agg(
        QTD_RES=("CD_RE", "nunique")
    )
    .sort_values("QTD_RES", ascending=False)
)

res_por_gv_df
```

```python
gvs_por_re_df = (
    zone_vs_setor_df
    .groupby("CD_RE", as_index=False)
    .agg(
        QTD_GVS=("CD_GV", "nunique")
    )
    .sort_values("QTD_GVS", ascending=False)
)

gvs_por_re_df
```

```python
gl_df = pd.read_excel("Bases_setores_cd.xlsx", sheet_name = "GI")
```

```python
gl_df
```

```python
zoneamentos_por_centro_df = (
    gl_df
    .groupby("CENTRO", as_index=False)
    .agg(
        QTD_ZONEAMENTOS=("ZONEAMENTO PARTIDA", "nunique")
    )
    .sort_values("QTD_ZONEAMENTOS", ascending=False)
)

zoneamentos_por_centro_df
```

```python
centros_por_zoneamento_df = (
    gl_df
    .groupby("ZONEAMENTO PARTIDA", as_index=False)
    .agg(
        QTD_CENTROS=("CENTRO", "nunique")
    )
    .sort_values("QTD_CENTROS", ascending=False)
)

centros_por_zoneamento_df
```
